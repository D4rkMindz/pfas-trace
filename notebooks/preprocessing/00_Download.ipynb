```python
import requests
import pandas as pd
from pfas_trace.config import settings, DATASET_FILE
```

```python
if not DATASET_FILE.exists():
    with requests.get(settings.DATASET_URL, stream=True, timeout=60) as r:
        r.raise_for_status()
        with open(DATASET_FILE, 'wb') as f:
            for chunk in r.iter_content(chunk_size=8192):
                f.write(chunk)
```

```python
df_export = pd.read_parquet(DATASET_FILE)
df_export.shape
```
